B. Kueri Bertingkat (Read)

1. Menampilkan Kopi dengan Tingkat Kepahitan di Atas 3

```python
print("Menu Kopi dengan tingkat kepahitan di atas 3:")

for menu in koleksi_menu.find({
    "kategori": "Kopi",
    "tingkat_kepahitan": {"$gt": 3}
}):
    print(menu["nama_menu"], "-", menu["tingkat_kepahitan"])
```

2. Menampilkan Menu dengan Harga Rp15.000 sampai Rp30.000

```python
print("Menu dengan harga 15000 sampai 30000:")

for menu in koleksi_menu.find({
    "harga": {
        "$gte": 15000,
        "$lte": 30000
    }
}):
    print(menu["nama_menu"], "- Rp", menu["harga"])
```

3. Menampilkan Menu yang Tidak Memiliki Field pedas_level

```python
print("Menu yang tidak memiliki field pedas_level:")

for menu in koleksi_menu.find({
    "pedas_level": {"$exists": False}
}):
    print(menu["nama_menu"], "-", menu["kategori"])
```

C. Pembaruan Data (Update)

```python
koleksi_menu.update_many(
    {},
    {
        "$set": {
            "status": "Tersedia"
        }
    }
)

print("Status berhasil ditambahkan ke seluruh menu!")
```

2. Mengubah Status Dua Menu Menjadi Habis

```python
# Mengubah status Espresso
koleksi_menu.update_one(
    {"nama_menu": "Espresso"},
    {"$set": {"status": "Habis"}}
)

# Mengubah status Ayam Geprek
koleksi_menu.update_one(
    {"nama_menu": "Ayam Geprek"},
    {"$set": {"status": "Habis"}}
)

print("Status Espresso dan Ayam Geprek berhasil diubah!")
```

D. Integrasi ke Spark

1. Mengambil Data dari MongoDB

```python
import pandas as pd

# Mengambil seluruh data dari MongoDB
dokumen_menu = list(koleksi_menu.find({}))

# Mengubah data menjadi pandas DataFrame
df_pandas = pd.DataFrame(dokumen_menu)

# Mengubah ObjectId menjadi string
df_pandas["_id"] = df_pandas["_id"].astype(str)

print("Data berhasil diambil dari MongoDB!")
print(df_pandas)
```

2. Mengubah Data Menjadi Spark DataFrame

```python
from pyspark.sql import SparkSession
from pyspark.sql.types import (
    StructType,
    StructField,
    StringType,
    IntegerType
)

# Membuat SparkSession
spark = SparkSession.builder \
    .appName("KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()

spark.sparkContext.setLogLevel("ERROR")

# Menentukan struktur kolom
schema = StructType([
    StructField("_id", StringType(), True),
    StructField("nama_menu", StringType(), True),
    StructField("harga", IntegerType(), True),
    StructField("kategori", StringType(), True),
    StructField("tingkat_kepahitan", IntegerType(), True),
    StructField("pedas_level", IntegerType(), True),
    StructField("status", StringType(), True)
])

# Mengambil data dan menyesuaikan field yang tidak tersedia
data_spark = []

for menu in dokumen_menu:
    data_spark.append((
        str(menu["_id"]),
        menu.get("nama_menu"),
        menu.get("harga"),
        menu.get("kategori"),
        menu.get("tingkat_kepahitan"),
        menu.get("pedas_level"),
        menu.get("status")
    ))

# Membuat Spark DataFrame
df_spark = spark.createDataFrame(data_spark, schema)

print("Spark DataFrame berhasil dibuat!")
```

3. Menampilkan Struktur Data

```python
df_spark.printSchema()
```

4. Menampilkan Data Spark DataFrame

```python
df_spark.show(truncate=False)
```

5. Menghitung Rata-Rata Harga Berdasarkan Kategori

```python
from pyspark.sql.functions import avg

hasil_analisis = df_spark.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga")
)

hasil_analisis.show()
```

E. Kesimpulan

Setelah kita mengerjakan tugas ini, kita jadi tahu kalau MongoDB bisa nyimpen data kopi dan makanan meskipun isi datanya beda. Contohnya, kopi punya tingkat kepahitan, sementara makanan punya pedas_level. Nah, masalah mulai muncul waktu data dari MongoDB mau diolah di PySpark, karena PySpark butuh susunan data yang jelas. Jadi, aku harus nyesuaiin kolomnya dulu dan ngasih nilai null buat bagian yang kosong. Di sini aku menggunakan StructType dan StructField supaya datanya bisa dibaca dengan benar. Setelah itu, data berhasil ditampilkan dan rata-rata harga tiap kategori juga bisa dihitung. Tantangannya ada di penyesuaian bentuk data tadi, karena kalau susunannya nggak sesuai, data bisa gagal dibaca oleh PySpark.
